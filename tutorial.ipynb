# Task 4 (optional): Run the entire pipeline in lazy mode

```python
# Eager Execution
raw_eager = RawData(models, policies)
start = time.perf_counter()
report_eager = run_pipeline(raw_eager)
report_eager.to_string()
eager_time = time.perf_counter() - start
print(f"Eager time: {eager_time:.4f}s")

# Lazy Execution
raw_lazy = RawData(models.lazy(), policies.lazy())
start = time.perf_counter()
report_lazy = run_pipeline(raw_lazy)
report_lazy.to_string()
lazy_time = time.perf_counter() - start
print(f"Lazy time: {lazy_time:.4f}s")

print(f"Speedup: {eager_time / lazy_time:.2f}x")
```

```python
# Show unoptimized graph
report_lazy.popularity.show_graph(optimized=False)
```

```python
# Show optimized graph
report_lazy.popularity.show_graph(optimized=True)
```
